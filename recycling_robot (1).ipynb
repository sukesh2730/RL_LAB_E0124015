# Recycling Robot

This notebook implements a simple recycling robot that sorts items by material and separates contaminated waste.

```python
class RecyclingRobot:
    def __init__(self):
        self.sorted_items = {
            'paper': 0,
            'plastic': 0,
            'metal': 0,
            'glass': 0,
            'contaminated': 0
        }

    def sort_item(self, material, contaminated=False):
        if contaminated:
            self.sorted_items['contaminated'] += 1
            return 'contaminated'

        if material in self.sorted_items:
            self.sorted_items[material] += 1
            return material

        return 'unknown'

    def report(self):
        print('Sorted items summary:')
        for key, value in self.sorted_items.items():
            print(f'{key}: {value}')

robot = RecyclingRobot()
items = [
    ('paper', False),
    ('plastic', False),
    ('metal', True),
    ('glass', False),
    ('paper', False),
    ('plastic', True),
    ('metal', False),
]

for material, contaminated in items:
    result = robot.sort_item(material, contaminated)
    print(f'Item: {material}, contaminated={contaminated} -> {result}')

robot.report()
```
